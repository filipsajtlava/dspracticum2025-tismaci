# Merge pdf to text

```python
import os
import pdfplumber
```

```python
output_text_file = "complete_text.txt"

def pdf_to_text(parent_dir: str, output_text_file: str):
    folders = [
        name for name in os.listdir(parent_dir)
        if os.path.isdir(os.path.join(parent_dir, name))
    ]
    print(f"Found folders: {folders}\n")

    if output_text_file in os.listdir():
        print("Found a file with the same name - deleting...\n")
        os.remove(output_text_file)

    all_text = ""
    for folder_name in folders:
        skip_pages = int(folder_name)

        folder_dir = os.path.join(parent_dir,folder_name)
        print(f"======== Checking folder: '{folder_name}' ========")
        for file in os.listdir(folder_dir):
            input_pdf_path = os.path.join(folder_dir, file)
            print(f"Converting file {file}")
            
            all_text += "nazov knihy ==================" + file + "\n"
            try:
                with pdfplumber.open(input_pdf_path) as pdf:
                    for page_num in range(skip_pages, len(pdf.pages)-1):
                        page = pdf.pages[page_num] # pdfplumber accesses pages via the .pages list
                        text = page.extract_text()  # pdfplumber extraction
                        
                        if text: # Add a safety check in case extraction returns None
                            all_text += text
                            
                    print("File successfully converted")
                    
            except ValueError:
                print("File wasn't converted, a ValueError occured - skipping.")
            except Exception as e:
                print(f"File wasn't converted due to an unexpected error: {e} ")
            
            print("--------")
        print()

    with open(output_text_file, "w", encoding="utf-8") as f:
        f.write(all_text)
```

```python
pdf_to_text(parent_dir="text_in", output_text_file="complete_text.txt")
```

```python
import re

def clean_text(text):
    # Keep only letters, digits, spaces, punctuation you choose
    allowed_pattern = r"[^a-zA-Z0-9\s.,!?–—…()=:;/'\"]"
    cleaned = re.sub(allowed_pattern, "", text)
    return cleaned

# --- Step 1: Read your .txt file ---
input_path = "complete_text.txt"
output_path = "cleaned_text.txt"
```

```python
with open(input_path, "r", encoding="utf-8", errors="ignore") as f:
    raw_text = f.read()

cleaned_text = clean_text(raw_text)
with open(output_path, "w", encoding="utf-8") as f:
    f.write(cleaned_text)
```
